```python
import pandas as pd
from pathlib import Path

data_file = Path("..") / "data" / "heart+disease" / "processed.cleveland.data"
column_names = [
    "age","sex","cp","trestbps","chol","fbs","restecg",
    "thalach","exang","oldpeak","slope","ca","thal","target",
]

df = pd.read_csv(data_file, header=None, names=column_names, na_values='?')

df.shape
df.head()
```

```python
print("Dimensões do dataset:")
print(df.shape)

print("\nPrimeiras 5 linhas:")
display(df.head())

print("\nTipos das colunas:")
print(df.dtypes)

print("\nValores ausentes:")
print(df.isnull().sum())
```

```python
print("Distribuição do target:")
print(df["target"].value_counts().sort_index())
```

```python
print("Estatísticas descritivas:")
display(df.describe())
```

```python
print("Distribuição do target:")
print(df["target"].value_counts().sort_index())
```

```python
df["target_binary"] = (df["target"] > 0).astype(int)

print("Distribuição do target binário:")
print(df["target_binary"].value_counts().sort_index())
```

```python
print("Registros com valores ausentes:")
display(df[df.isnull().any(axis=1)])
```

```python
print("Moda de ca:", df["ca"].mode()[0])
print("Moda de thal:", df["thal"].mode()[0])
```

```python
for coluna in df.columns:
    print(f"{coluna}: {df[coluna].nunique()} valores únicos")
```

```python
import matplotlib.pyplot as plt

df[["age", "trestbps", "chol", "thalach", "oldpeak"]].hist(
    figsize=(12, 8),
    bins=20
)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

df.boxplot(column="age", by="target_binary")

plt.title("Distribuição da idade por presença de doença")
plt.suptitle("")
plt.xlabel("Doença cardíaca (0 = não, 1 = sim)")
plt.ylabel("Idade")

plt.show()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(14, 8))

variaveis = ["age", "trestbps", "chol", "thalach", "oldpeak"]

for ax, coluna in zip(axes.flat, variaveis):
    df.boxplot(column=coluna, by="target_binary", ax=ax)
    ax.set_title(coluna)
    ax.set_xlabel("Doença (0 = não, 1 = sim)")
    ax.set_ylabel(coluna)

axes[1, 2].axis("off")

plt.suptitle("Variáveis numéricas por presença de doença")
plt.tight_layout()
plt.show()
```

```python
pd.crosstab(
    df["sex"],
    df["target_binary"],
    normalize="index"
) * 100
```

```python
variaveis_cat = [
    "sex", "cp", "fbs", "restecg",
    "exang", "slope", "ca", "thal"
]

fig, axes = plt.subplots(2, 4, figsize=(16, 8))

for ax, coluna in zip(axes.flat, variaveis_cat):
    pd.crosstab(
        df[coluna],
        df["target_binary"],
        normalize="index"
    ).plot(
        kind="bar",
        stacked=True,
        ax=ax,
        legend=False
    )
    
    ax.set_title(coluna)
    ax.set_xlabel("")
    ax.set_ylabel("%")
    ax.tick_params(axis="x", rotation=0)

plt.suptitle("Variáveis categóricas por presença de doença")
plt.tight_layout()
plt.show()
```
